```python
# Question 1
def strings_longer_than_k(strings, k):
    result = []
    for text in strings:
        if len(text) > k:
            result.append(text)
    return result


words = ["apple", "pie", "banana"]
print(strings_longer_than_k(words, 3))
```

```python
# Question 2
def area(shape, *dimensions):
    shape = shape.lower()

    if shape == "circle" and len(dimensions) == 1:
        radius = dimensions[0]
        return 3.14 * radius ** 2
    elif shape == "rectangle" and len(dimensions) == 2:
        length, width = dimensions
        return length * width
    else:
        return "Invalid shape or dimensions"


print("Circle Area:", area("circle", 5))
print("Rectangle Area:", area("rectangle", 4, 6))
```

```python
# Question 3
def split_email(email):
    if "@" not in email:
        return None, None

    username, domain = email.split("@", 1)
    return username, domain


email = "support@google.com"
username, domain = split_email(email)
print("Username:", username)
print("Domain:", domain)
```

```python
# Question 4
def sum_numbers(values):
    total = 0
    for value in values:
        if isinstance(value, (int, float)) and not isinstance(value, bool):
            total += value
    return total


values = [10, "apple", 20, True, 5]
print("Total Sum:", sum_numbers(values))
```

## Question 5


```python
# Question 5
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer_name, table_number):
        reservation = {"customer": customer_name, "table": table_number}
        self.book_table.append(reservation)

    def customer_order(self, customer_name, items):
        order = {"customer": customer_name, "items": items}
        self.customer_orders.append(order)

    def print_menu(self):
        print("Menu:")
        for item, price in self.menu_items.items():
            print(f"{item}: {price}")

    def print_table_reservations(self):
        print("Table Reservations:")
        for reservation in self.book_table:
            print(f"{reservation['customer']} - Table {reservation['table']}")

    def print_customer_orders(self):
        print("Customer Orders:")
        for order in self.customer_orders:
            print(f"{order['customer']}: {', '.join(order['items'])}")


restaurant = Restaurant()
restaurant.add_item_to_menu("Pizza", 120)
restaurant.add_item_to_menu("Burger", 80)
restaurant.add_item_to_menu("Juice", 30)

restaurant.book_tables("Ahmed", 5)
restaurant.book_tables("Sara", 2)

restaurant.customer_order("Ahmed", ["Pizza", "Juice"])
restaurant.customer_order("Sara", ["Burger"])

restaurant.print_menu()
restaurant.print_table_reservations()
restaurant.print_customer_orders()
```

```python
# Question 6
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("The customer must be over 6 years old.")
            return []

        if number_of_bikes <= 0:
            print("Please request at least one bike.")
            return []

        if number_of_bikes > len(self.bikes):
            print("Requested bikes are not available. Please wait 10 minutes.")
            return []

        rented_bikes = []
        for _ in range(number_of_bikes):
            rented_bikes.append(self.bikes.pop())

        rental = {"customer": customer, "bikes": rented_bikes}
        self.rentals.append(rental)
        print(f"{customer.get_name()} rented {number_of_bikes} bike(s).")
        return rented_bikes

    def returnBike(self, rentalTime, num_of_bikes):
        if num_of_bikes <= 0:
            print("The number of returned bikes must be greater than zero.")
            return 0

        bikes_to_return = []
        while self.rentals and len(bikes_to_return) < num_of_bikes:
            rental = self.rentals[-1]
            while rental["bikes"] and len(bikes_to_return) < num_of_bikes:
                bikes_to_return.append(rental["bikes"].pop())
            if not rental["bikes"]:
                self.rentals.pop()

        if len(bikes_to_return) < num_of_bikes:
            print("There are not enough rented bikes to return.")

        self.bikes.extend(bikes_to_return)
        total_bill = rentalTime * 40
        print(f"Returned {len(bikes_to_return)} bike(s). Total bill: {total_bill}")
        return total_bill

    def addBike(self, bike):
        self.bikes.append(bike)

    def totalCost(self):
        return sum(bike.get_price() for bike in self.bikes)


bike_shop = MainBikeRental()
bike_shop.addBike(Bike("Mountain Bike", 5000))
bike_shop.addBike(Bike("Road Bike", 6500))
bike_shop.addBike(Bike("City Bike", 4000))

customer = Customer("Omar", 20)
print("Available bikes:", len(bike_shop.get_bikes()))
print("Total cost of bikes in the store:", bike_shop.totalCost())
bike_shop.requestBike(customer, 2)
print("Available bikes after rental:", len(bike_shop.get_bikes()))
bike_shop.returnBike(3, 2)
print("Available bikes after return:", len(bike_shop.get_bikes()))
```
