```python
import numpy as np

# Reproducible random number generator
rng = np.random.default_rng(42)
```

### 1. Min-Max Normalization (3x3 Array)
Create a 3x3 array with random integers between 1 and 100. Write a program to "normalize" the array so that the values are scaled between 0 and 1.
- **Formula:** $\frac{x - \min}{\max - \min}$
- **Example Output:** An array where the smallest value becomes `0.0` and the largest becomes `1.0`.

```python
arr_q1 = rng.integers(1, 101, size=(3, 3))

min_val = arr_q1.min()
max_val = arr_q1.max()

if max_val == min_val:
    normalized_q1 = np.zeros_like(arr_q1, dtype=float)
else:
    normalized_q1 = (arr_q1 - min_val) / (max_val - min_val)

print("Original 3x3 Array:\n", arr_q1)
print(f"\nMin: {min_val}, Max: {max_val}")
print("\nNormalized Array (0.0 to 1.0):\n", np.round(normalized_q1, 4))
```

### 2. Closest Value to X in a 1D Array
Create a 1D array of 10 random numbers. Ask the user for a value `X`. Find the value in the array that is closest to `X`.
- **Example Input:** Array: `[1, 5, 8, 12]`, `X = 10`
- **Example Output:** `8` or `12` (whichever is mathematically closer).

```python
arr_q2 = rng.integers(1, 51, size=10)
print("1D Array:", arr_q2)

x = float(input("Enter a value for X: "))

# Compute absolute distance to X and find index of minimum distance
closest_idx = np.abs(arr_q2 - x).argmin()
closest_val = arr_q2[closest_idx]

print(f"Closest value to {x} in the array: {closest_val}")
```

### 3. Second-Largest Value per Column (5x5 Array)
Write a NumPy program to create a 5x5 array with random values. For each column in the array, find the second-largest value.
- **Example Input:** `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]`
- **Example Output:** `[4, 5, 6]`

```python
arr_q3 = rng.integers(1, 101, size=(5, 5))

# Sort each column ascending along axis=0, then pick the second element from the end (-2)
sorted_cols = np.sort(arr_q3, axis=0)
second_largest_per_col = sorted_cols[-2, :]

print("5x5 Array:\n", arr_q3)
print("\nSecond-largest value per column:\n", second_largest_per_col)
```

### 4. Row Indices with Even Row Sums (6x4 Matrix)
Create a 6x4 matrix of random integers. Identify which rows have a sum that is an even number. Print the indices of these rows.

```python
arr_q4 = rng.integers(1, 51, size=(6, 4))
row_sums = arr_q4.sum(axis=1)
even_row_indices = np.where(row_sums % 2 == 0)[0]

print("6x4 Matrix:\n", arr_q4)
print("\nRow sums:\n", row_sums)
print("\nIndices of rows with even sums:", even_row_indices)
```

### 5. Replace Multiples of 3 or 5 with -1 (1D Array of 30)
Create a 1D array of 30 random integers between 1 and 50. Replace all numbers that are multiples of 3 or 5 with the value `-1`.
- **Example Input:** `[3, 7, 10, 15]`
- **Example Output:** `[-1, 7, -1, -1]`

```python
arr_q5 = rng.integers(1, 51, size=30)
result_q5 = arr_q5.copy()

mask = (result_q5 % 3 == 0) | (result_q5 % 5 == 0)
result_q5[mask] = -1

print("Original Array:\n", arr_q5)
print("\nModified Array (multiples of 3 or 5 -> -1):\n", result_q5)
```

### 6. Row-Wise Mean Centering (3x3 Array)
Write a NumPy program to create a 3x3 array with random values. For each row, calculate its mean (average) and subtract that mean from every element in that specific row.
- **Example Input:** `[[10, 20, 30], [5, 5, 5], [0, 10, 20]]`
- **Example Output:** `[[-10., 0., 10.], [0., 0., 0.], [-10., 0., 10.]]`

```python
arr_q6 = rng.integers(1, 51, size=(3, 3)).astype(float)

# keepdims=True preserves shape (3, 1) so broadcasting subtracts each row's mean cleanly
row_means = arr_q6.mean(axis=1, keepdims=True)
centered_q6 = arr_q6 - row_means

print("Original 3x3 Array:\n", arr_q6)
print("\nRow Means:\n", row_means.ravel())
print("\nMean-Centered Array:\n", np.round(centered_q6, 4))
```
