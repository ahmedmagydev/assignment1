## Q1. Separate letters and numbers


```python
text = "Py2th0on3"
# text = input("Enter a string with letters and numbers: ")

letters = []
numbers = []
for ch in text:
    if ch.isalpha():
        letters.append(ch)
    elif ch.isdigit():
        numbers.append(int(ch))

word = "".join(letters)
print(f'Word: "{word}", Sum: {sum(numbers)}')
```

## Q2. Password validation
Valid if: length 8–16, at least one uppercase letter, and at least one special character.

```python
password = "Guest@123"
# password = input("Enter a password: ")

special_chars = "!@#$%^&*()-_=+[]{};:'\",.<>/?\\|`~"

length_ok = 8 <= len(password) <= 16
has_upper = any(ch.isupper() for ch in password)
has_special = any(ch in special_chars for ch in password)

if length_ok and has_upper and has_special:
    print("Password is Secure.")
else:
    print("Password is Not Secure.")
    if not length_ok:
        print("- Length must be between 8 and 16 characters.")
    if not has_upper:
        print("- Must contain at least one uppercase letter.")
    if not has_special:
        print("- Must contain at least one special character.")
```

## Q3. Store discount
Under \$50: no discount · \$50–\$100: 10% · over \$100: 20%. Members get an extra 5% off the result.

```python
price = 100
member = "Yes"
# price = float(input("Price: "))
# member = input("Member (Yes/No): ")

if price < 50:
    total = price
elif price <= 100:
    total = price * 0.90
else:
    total = price * 0.80

if member.strip().lower() in ("yes", "y"):
    total = total * 0.95

print(f"Final Total: ${round(total, 2)}")
```

## Q4. Common names in two lists

```python
list1_text = "Ali, Ben, Sara"
list2_text = "Ben, Joy, Ali"
# list1_text = input("List 1 (comma-separated): ")
# list2_text = input("List 2 (comma-separated): ")

list1 = [name.strip() for name in list1_text.split(",")]
list2 = [name.strip() for name in list2_text.split(",")]

common = []
for name in list1:
    if name in list2 and name not in common:
        common.append(name)

print("Common names:", sorted(common))
```

## Q5. Student averages — Honor Roll
Print "Honor Roll" when a student's average is above 90.

```python
students = {
    "Sam": [95, 90, 92],
    "Lina": [80, 85, 78],
    "Omar": [99, 94, 91],
}

for name, scores in students.items():
    average = sum(scores) / len(scores)
    if average > 90:
        print(f"{name}: Average {average:.1f} - Honor Roll")
    else:
        print(f"{name}: Average {average:.1f}")
```

## Q7. Digital root
Keep summing the digits until a single digit is left.

```python
number = "98"
# number = input("Enter a multi-digit number: ").strip()

steps = []
while len(number) > 1:
    total = sum(int(d) for d in number)
    steps.append("+".join(number) + f"={total}")
    number = str(total)

print(f"Digital Root: {number} ({', '.join(steps)})")
```

## Q9. Factorial and divisibility
Compute n! with a loop, then check if it is divisible by 1 + 2 + … + n.

```python
n = 4
# n = int(input("Enter n: "))

factorial = 1
total = 0
for i in range(1, n + 1):
    factorial *= i
    total += i

if total != 0 and factorial % total == 0:
    print(f"Factorial is {factorial}, Sum is {total}. {factorial} is divisible by {total}.")
else:
    print(f"Factorial is {factorial}, Sum is {total}. {factorial} is not divisible by {total}.")
```

## Q10. Shopping cart
Commands are a list here so the cell runs on its own. Set `interactive = True` to type commands yourself (`add apple`, `remove milk`, `done`).

```python
store = {"apple": 0.5, "bread": 2.0, "milk": 1.5}
cart = []

interactive = False
commands = ["add apple", "add milk", "add pizza"]   # example input


def handle(command):
    parts = command.strip().lower().split(maxsplit=1)
    if len(parts) != 2:
        print("Invalid command.")
        return
    action, item = parts[0], parts[1].strip("'\" ")
    if action == "add":
        if item in store:
            cart.append(item)
            print(f"Added {item}.")
        else:
            print(f"Error: '{item}' is not in the store.")
    elif action == "remove":
        if item in cart:
            cart.remove(item)
            print(f"Removed {item}.")
        else:
            print(f"Error: '{item}' is not in your cart.")
    else:
        print("Invalid command.")


if interactive:
    while True:
        command = input("> ")
        if command.strip().lower() == "done":
            break
        handle(command)
else:
    for command in commands:
        handle(command)

total = sum(store[item] for item in cart)
print(f"Total Bill: ${total}")
```

## Q13. Find peaks
A peak is greater than both its left and right neighbours.

```python
numbers = [1, 5, 2, 6, 3]

peaks = []
for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
```

## Q14. Flip a dictionary
Keys that share a value are grouped in a list.

```python
original = {"A": 1, "B": 2, "C": 1}

flipped = {}
for key, value in original.items():
    if value not in flipped:
        flipped[value] = []
    flipped[value].append(key)

print(flipped)
```

## Q15. Vowel indices

```python
text = "banana"
# text = input("Enter a string: ")

result = {}
for index, ch in enumerate(text):
    if ch.lower() in "aeiou":
        result.setdefault(ch.lower(), []).append(index)

print(result)
```
